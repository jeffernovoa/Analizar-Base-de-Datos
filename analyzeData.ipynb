# Análisis de una Base de Datos

```python
import matplotlib.pyplot as plt
import seaborn as sns

import pandas as pd

df = pd.read_csv('clientes.csv')
columns = df.columns

print("Tamaño de la DB:", df.shape)
print("Columnas de la DB:", df.columns)
print("Tipos de datos:", df.dtypes)
print(df.head())
```

# Revisión de tipos de datos

```python
for column in columns:
    values = df[column].value_counts()
    print(values)
    print()
```

```python
df.info()
```

# Creación de histogramas

```python
x = df['cuota_mensual'].values

x_mean = x.mean()

plt.hist(x-x_mean, bins=40)
plt.show()
```

```python
X0 = df[df['abandono'] == 'No']
X1 = df[df['abandono'] == 'Sí']

for col in ['edad', 'ingresos_mensuales', 'cuota_mensual', 'satisfaccion']:
    plt.hist(X0[col].values, bins = 50, alpha=0.5, label='No abandono')
    plt.hist(X1[col].values, bins = 50, alpha=0.5, label='Abandono')
    plt.legend()
    plt.show()
```
